```python
import yfinance as yf
import pandas as pd

tickers = ['AAPL', 'JPM', 'XOM', 'SPY', 'NVDA']
data = yf.download(tickers, start='2025-01-01', end='2026-01-01', auto_adjust=False)

for ticker in tickers: 
    comparison = data[[("Close", ticker), ("Adj Close", ticker)]].copy()
    comparison.columns = ['Close', 'Adj Close']

    comparison['Difference'] = (comparison['Close'] - comparison['Adj Close']) / comparison['Close'] * 100

    print(f"\n==== {ticker} ====")
    print(comparison.head(10))
```
